```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.load("data/2026-05-07/#588_resonator_spectroscopy_111238/arrays.npz")

I = data['I_data']
Q = data['Q_data']
Freq = data['frequency']
Amp = np.sqrt(I*I + Q*Q)

idx_res = np.argmin(Amp)

S21 = I + 1j*Q
S21 = S21 / np.abs(S21[0])

phase = np.unwrap(np.angle(S21))

mask = np.ones_like(Freq, dtype=bool)

f0_guess = Freq[np.argmin(np.abs(S21))]
bw = 1e6
mask &= np.abs(Freq - f0_guess) > bw

p = np.polyfit(Freq[mask], phase[mask], 1)
phase_bg = np.polyval(p, Freq)

S21_corr = S21 * np.exp(-1j * phase_bg)

fig, ax = plt.subplots(1, 3, figsize=(12, 4))

ax[0].plot(Freq / 1e9, Amp)
ax[0].axvline(x=Freq[idx_res] / 1e9, color='red')
ax[0].set_xlabel('Frequency (GHz)')
ax[0].set_ylabel(r'$\sqrt{I^2 + Q^2}$ (mV)')
ax[0].set_title('Resonator Spectrum')

ax[1].scatter(I, Q, s=2)
ax[1].scatter(I[idx_res], Q[idx_res], marker='x', color='red')
ax[1].set_xlabel('I (mV)')
ax[1].set_ylabel('Q (mV)')
ax[1].set_title('I/Q Plot')

ax[2].plot(np.real(S21_corr), np.imag(S21_corr), zorder=1)
ax[2].scatter(np.real(S21_corr[idx_res]), np.imag(S21_corr[idx_res]), marker='x', color='red', zorder=2)
ax[2].spines['bottom'].set_position('zero')
ax[2].spines['top'].set_visible(False)
ax[2].spines['right'].set_visible(False)

ax[2].set_xlim(0.0, 1.0)
ax[2].set_ylim(-0.5, 0.5)

ax[2].set_xlabel('Real').set_zorder(4)
ax[2].set_ylabel('Imaginary').set_zorder(4)
ax[2].set_title('Normalized and De-embedded')

plt.tight_layout()
plt.show()
```
