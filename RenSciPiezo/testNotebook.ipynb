# Low level TCP/IP communication with the JSON-RPC server on the RenSci driver

```python
def sendRPC(client, rpc):
    rpc_bytes = json.dumps(rpc).encode('utf-8') # Convert the call to a byte array.
    try: # Send the command and gather the response.
        client.send(rpc_bytes)
        response = client.recv(128)
        print(rpc_bytes)
    except socket.error:
        print('Failed to send data')
    response_json = json.loads(response) # Convert the response back to JSON format
    # for easy manipulation.
    return response_json
```

```python
with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as client:
    client.connect((HOST, PORT))
    print(f'Socket Connected to {HOST}:{PORT}')
    rpc = {
        "method": "goStepsForward", # Method name. goStepsForward(channel, steps, amplitude, frequency)
        "params": ["2", "100", "200", "1000"], # Parameters.
        "jsonrpc": "2.0", # JSON-RPC spec, only 2.0 is supported.
        "id": 0, # ID of procedure call, useful for
        } # checking the result.
    resp = sendRPC(client, rpc)
    print(f"id# {resp['id']} --- Return Value: {str(resp['result'])}")
```
